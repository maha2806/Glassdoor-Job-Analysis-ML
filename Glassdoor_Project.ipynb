### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

Based on exploratory data analysis and visualization, three hypothetical statements were formulated to statistically examine relationships between variables in the Glassdoor Jobs dataset. Statistical testing helps determine whether observed patterns are significant or due to random variation. Hypothesis testing was performed using p-values and standard significance levels.

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Null Hypothesis (H₀):
There is no significant relationship between company ratings and company founded year.

Alternative Hypothesis (H₁):
There is a significant relationship between company ratings and company founded year.

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value
````

````python
from scipy.stats import pearsonr

rating = df['Rating']
founded = df['Founded']

correlation,p_value = pearsonr(
rating,
founded
)

print("P-value:",p_value)
````

##### Which statistical test have you done to obtain P-Value?

Pearson Correlation Test

##### Why did you choose the specific statistical test?

Pearson correlation was selected because both variables are numerical and the objective is to identify whether a relationship exists between them.

Conclusion:
If p-value < 0.05, reject the null hypothesis. Otherwise, fail to reject the null hypothesis.

### Hypothetical Statement - 2

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Null Hypothesis (H₀):
The average company ratings are the same across different industries.

Alternative Hypothesis (H₁):
The average company ratings differ across industries.

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value
````

````python
from scipy.stats import f_oneway

groups=[]

for industry in df['Industry'].unique()[:5]:

    groups.append(
        df[df['Industry']==industry]
        ['Rating']
    )

f_stat,p_value=f_oneway(*groups)

print("P-value:",p_value)
````

##### Which statistical test have you done to obtain P-Value?

ANOVA (Analysis of Variance)

##### Why did you choose the specific statistical test?

ANOVA is suitable because multiple groups (industries) are being compared to determine whether their mean ratings differ significantly.


Conclusion:
If p-value < 0.05, reject the null hypothesis.

### 1. Handling Missing Values

````python
# Handling Missing Values & Missing Value Imputation
````

````python
# Handling Missing Values

print(df.isnull().sum())

# Replace missing rating values
df['Rating'].fillna(
    df['Rating'].median(),
    inplace=True
)

print(df.isnull().sum())
````

#### What all missing value imputation techniques have you used and why did you use those techniques?

Median imputation was used to handle missing values in the Rating column. Median was selected because it reduces the impact of extreme values and preserves the central tendency of the dataset. This technique ensures that missing values are replaced without significantly affecting the data distribution.

### 2. Handling Outliers

````python
# Handling Outliers & Outlier treatments
````

````python
# Outlier detection

plt.figure(figsize=(8,5))

sns.boxplot(
x=df['Rating']
)

plt.show()
````

````python
Q1=df['Rating'].quantile(0.25)

Q3=df['Rating'].quantile(0.75)

IQR=Q3-Q1

lower=Q1-(1.5*IQR)
upper=Q3+(1.5*IQR)

df=df[
(df['Rating']>=lower)
&
(df['Rating']<=upper)
]
````

##### What all outlier treatment techniques have you used and why did you use those techniques?

The Interquartile Range (IQR) method was used for outlier detection and treatment. IQR was chosen because it effectively identifies values outside the normal data spread and reduces the influence of extreme observations.

````python
# Remove White spaces

df['Job Description'] = df['Job Description'].apply(
    lambda x: " ".join(x.split())
)

df['Job Description'].head()
````

````python
# Rephrase Text

df['Job Description'] = df['Job Description'].apply(
    lambda x: str(x).strip()
)

df['Job Description'].head()
````

#### 7. Tokenization

````python
# Tokenization
````

````python
nltk.download('punkt')
nltk.download('punkt_tab')

from nltk.tokenize import word_tokenize

df['tokens']=df[
'Job Description'
].apply(word_tokenize)
````

#### 8. Text Normalization

````python
# Normalizing Text (i.e., Stemming, Lemmatization etc.)
````

````python
nltk.download('wordnet')

from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize # Ensure word_tokenize is available

lemmatizer=WordNetLemmatizer()

# Ensure 'tokens' column exists by re-tokenizing 'Job Description' if needed
# This step re-creates the 'tokens' column from the 'Job Description' if it was not created
# or was lost from the previous cell's execution, resolving the KeyError.
if 'tokens' not in df.columns:
    df['tokens'] = df['Job Description'].apply(word_tokenize)

df['normalized']=df[
'tokens'
].apply(
lambda words:[
lemmatizer.lemmatize(word)
for word in words
]
)
````

##### Which text normalization technique have you used and why?

Lemmatization was used because it converts words into meaningful root forms while preserving their context and meaning.

#### 9. Part of speech tagging

````python
# POS Taging
````

````python
# POS Tagging

import nltk

nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng') # Added to resolve LookupError

from nltk import pos_tag
from nltk.tokenize import word_tokenize

sample_text = str(df['Job Description'].iloc[0])

tokens = word_tokenize(sample_text)

pos_tags = pos_tag(tokens)

print(pos_tags[:20])
````

#### 10. Text Vectorization

````python
# Vectorizing Text
````

````python
from sklearn.feature_extraction.text import TfidfVectorizer

# Join the normalized tokens back into strings for TF-IDF vectorization
df['normalized_text'] = df['normalized'].apply(lambda x: ' '.join(x))

# Initialize TF-IDF Vectorizer
tfidf_vectorizer = TfidfVectorizer(max_features=5000) # Limiting to 5000 features for manageable size

# Fit and transform the 'normalized_text' column
tfidf_matrix = tfidf_vectorizer.fit_transform(df['normalized_text'])

# Convert the TF-IDF matrix to a DataFrame for easier inspection
tfidf_df = pd.DataFrame(tfidf_matrix.toarray(), columns=tfidf_vectorizer.get_feature_names_out())

print("TF-IDF Vectorization complete. Shape of TF-IDF matrix:", tfidf_matrix.shape)
print("First 5 rows of TF-IDF DataFrame:")
print(tfidf_df.head())
````

##### Which text vectorization technique have you used and why?

TF-IDF (Term Frequency-Inverse Document Frequency) was chosen for text vectorization. This technique reflects how important a word is to a document in a collection or corpus. It increases proportionally to the number of times a word appears in the document but is offset by the frequency of the word in the corpus, which helps to adjust for the fact that some words appear more frequently in general. This method is effective because it assigns higher weights to terms that are rare but highly relevant to a specific document, while downplaying common words like 'the' or 'a' that appear in many documents and thus carry less unique information. I have limited the maximum features to 5000 to manage dimensionality.

TF-IDF (Term Frequency–Inverse Document Frequency) was used because it converts textual data into numerical values while assigning importance to meaningful terms and reducing the effect of commonly occurring words.
